```python
import pandas as pd
import torch
import matplotlib.pyplot as plt
```

```python
# Crossfires 
csv_prn = pd.read_csv('tmps/curve_prn_1.csv')
csv_enR = pd.read_csv('tmps/curve_enR_1.csv')
csv_enL = pd.read_csv('tmps/curve_enL_1.csv')
save_dir=f'tmps/aela_crossfire_enL_classifier.png'

radii = torch.linspace(0, 40, len(csv_enR))
print("Plotting AELA figure...")
upper_limit = [3 / 4 * r for r in radii]
# Plotting the results
plt.figure(figsize=(10, 6))
plt.plot(radii, csv_prn, label='prn')
plt.plot(radii, csv_enR, label='enR')
plt.plot(radii, csv_enL, label='enL')
plt.plot(radii, upper_limit,  linestyle='--', color='blue', label='Upper Limit')
plt.legend(loc = 'best')
plt.xlabel('Radius (mm)')   
plt.ylabel('Average Expected Local Accuracy (AELA)')
plt.yscale('log')
# plt.xscale('log')
# plt.xlim(0, 15)
plt.grid()
plt.savefig(save_dir)
plt.show()
# Save the plot as a PNG file
#plt.close()  # Close the plot to free up memory
# Return the average expected local accuracy for each radius
```
